```python
T = 1
for _ in range(T):
    S = input().strip()
    stack=[]
    for s in S:
        if len(stack)==0:stack.append(s)
        elif s == 'o':
            if stack[-1]=='o':stack[-1] = 'O'
            else: stack.append('o')
        else:
            if stack[-1] == 'O':stack.pop()
    print(''.join(stack))
```
